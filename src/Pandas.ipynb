```python
import pandas as pd 
```

```python
data = [
    ['Nissan', 'Stanza', 1991, 138, 4, 'MANUAL', 'sedan', 2000],
    ['Hyundai', 'Sonata', 2017, None, 4, 'AUTOMATIC', 'Sedan', 27150],
    ['Lotus', 'Elise', 2010, 218, 4, 'MANUAL', 'convertible', 54990],
    ['GMC', 'Acadia',  2017, 194, 4, 'AUTOMATIC', '4dr SUV', 34450],
    ['Nissan', 'Frontier', 2017, 261, 6, 'MANUAL', 'Pickup', 32340],
]

columns = [
    'Make', 'Model', 'Year', 'Engine HP', 'Engine Cylinders',
    'Transmission Type', 'Vehicle_Style', 'MSRP'
]
```

```python
# creating a dataframe for the data or csv if attached
df = pd.DataFrame(data, columns=columns)
```

```python
# print the dataframe
df
```

```python
# we can columns like shown below. NOTE: if the there are 10 rows, we have to make sure there are 10 values appended or else it will throw an error.
df['id'] = [1,2,3,4,5]
```

```python
df
```

```python
# .index() will give the range of indcies of the dataframe
df.index
```

```python
# .loc[index] will return the rows. we will pass the array index (rows here) and it will return them. We can slicing techniques to get more personalized
# information
df.loc[3:]
```

```python
# we can change the index's datatype too. Here we are changing it from number to alphabet.
df.index = ['a', 'b','c','d','e']
```

```python
df
```

```python
# when we change the index from numbers to any other formats, we cannot use that .loc[] object. Pandas will have positional numbered index under the hood and we can 
# access it using .iloc[index]
df.iloc[[1,2]]
```

```python
# we can reset the indices back to its default by using reset_index() method. By default, it will store the legacy indices in a separate column. 
# If we don't need it, we can pass drop=True as an argument to the method.
df = df.reset_index(drop=True)
```

```python
df
```

```python
df['test'] = [10,20,30,40,50]
```

```python
df
```

```python
# We can multiply column values with a number, just like we do in numpy.
df['test'] * 20  #select a column and multiply with a number
```

```python
# we can so comparision operations.
df['Year'] >= 2012
```

```python
'''
Filtering operatoins

here we will be writing conditions inside the square brackets which is followed after df. example shown below. We can also write multiple conditions 
inside the square brackets.
'''

df[
    (df['Year'] >= 2015) & (df['Make'] == 'GMC')
]
```

# String methods

we can use string methods like str.lower() and str.uppper() and many more to perform string operations on the dataframe columns.m
Here we are going to convert the string attributes of the 'Vehicle type' column to lowercase.

```python
df['Vehicle_Style'].str.lower()
```

```python
# we can chain string operations as shown below. We can assign the operated column as a new replacement column to make sure we have the new series in the dataframe

df['Vehicle_Style'] = df['Vehicle_Style'].str.replace(' ', '_').str.upper()
```

```python
# In pandas, we can group by multiple columns, apply operations on them and view them just like we do it in SQL.

# for example, here we are going to group by the vehicles based on their 'transimission type' and 'MSRP' and find the max of 'MSRP'

df.groupby('Transmission Type').MSRP.max()
```
